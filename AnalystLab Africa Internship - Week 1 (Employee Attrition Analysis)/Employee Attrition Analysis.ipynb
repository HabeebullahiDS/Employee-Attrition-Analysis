2.**Sales (30.3%)** represents almost one-third of the workforce, making it the second-largest department. This suggests that sales and customer-related activities are also important but require fewer employees than Research & Development.

3.**Human Resources (4.3%**) has the smallest proportion of employees, representing only a small fraction of the total workforce. This is expected, as HR departments typically support employees rather than directly contribute to core business operations.

Overall, the organization’s workforce composition is dominated by Research & Development (65.4%), followed by Sales (30.3%), while Human Resources (4.3%) constitutes a relatively small support function. This distribution suggests that the company is primarily focused on technical and product-related activities.

# Part 4 – Initial Business Insights
Provide five workforce observations and five possible factors contributing to attrition supported by evidence.

### A. Five Workforce Observations

1. `The workforce is concentrated in the Research & Development department`.
Research & Development employs 961 employees (65.4%), making it the largest department in the organization. Sales accounts for 446 employees (30.3%), while Human Resources has only 63 employees (4.3%). This indicates that the organization primarily focuses on research and technical operations.

2. `Sales Executive is the most common job role`.
Out of the nine job roles, Sales Executive has the highest number of employees (326 employees; 22.2%), followed by Research Scientist (292 employees; 19.9%) and Laboratory Technician (259 employees; 17.6%). Together, these three roles represent 59.7% of the total workforce.

3. `Most employees do not work overtime`.
The majority of employees (1,054 employees; 71.7%) do not work overtime, while 416 employees (28.3%) regularly work overtime. This implies that  standard working hours are common and highly maintained across the organization.

4. `The workforce is predominantly composed of mid-career employees`.
Employee ages range from approximately 18 to 60 years, with the highest concentration between 25 and 40 years, specifically around `30–35 years`. This indicates that the organization mainly employs experienced, mid-career professionals.

5. `The organization has a relatively low overall employee attrition rate`.
Out of 1,470 employees, 237 employees (16.1%) have left the organization, while 1,233 employees (83.9%) have remained. This indicates that although the majority of employees stay with the organization,this indicates that employee turnover affected approximately 16 out of every 100 employees during the period covered by the dataset.

### B. Five Possible Factors Contributing to Employee Attrition

1. `Overtime appears to be the strongest contributor to employee attrition`.
Looking at the employees who work overtime, there was an attrition rate of 30.5%, compared to only 10.4% among employees who do not work overtime. This suggests that heavy workloads and extended working hours may significantly increase employees' likelihood of leaving the organization.

Evidence: Overtime attrition analysis (30.5% vs. 10.4%).

2. `Some job roles experience substantially higher employee turnover`.
Sales Representatives have the highest attrition rate (39.8%), followed by Laboratory Technicians (23.9%) and Human Resources (23.1%). In contrast, Research Directors (2.5%) and Managers (4.9%) have the lowest attrition rates. This can be due to high technical know-how and workload in those department with high attrition rate.

Evidence: Job Role vs. Attrition analysis.

3. `Younger employees are more likely to leave the organization`.
Employees who left have a median age of approximately 32 years, while those who stayed have a median age of approximately 36 years. This suggests that younger employees may be more likely to seek career advancement or alternative employment opportunities.

Evidence: Age by Attrition box plot.

4. `Lower monthly income is associated with higher attrition`.
Employees who left the organization have a median monthly income of approximately 3,200, compared to approximately 5,200 for employees who remained. This suggests that lower compensation may contribute to employee turnover.

Evidence: Monthly Income by Attrition box plot.

5. `Department influences employee attrition`.
The Sales department records the highest attrition rate (20.6%), followed by Human Resources (19.0%), while Research & Development has the lowest attrition rate (13.8%). This suggests that work environment, responsibilities, or departmental demands may influence employees' decisions to leave.

Evidence: Department vs. Attrition analysis.

### C. Business Recommendations

1. `Implement Work-Life Balance Initiatives`

The analysis indicates that employees who work overtime experience higher attrition rates than those who do not. To reduce employee burnout and improve job satisfaction, the organisation should regularly monitor overtime hours, encourage flexible working arrangements where appropriate, and ensure that workloads are distributed fairly among employees.


2. `Develop Department-Specific Retention Strategies`

Since attrition rates differ across departments, the organisation should prioritise departments with higher employee turnover for further investigation. Management should conduct employee engagement surveys, review leadership practices, assess workload distribution, and provide targeted interventions to address department-specific challenges contributing to employee attrition.

3. `Strengthen Career Development and Employee Growth Opportunities`

The variation in attrition across job roles suggests that career progression and professional development may influence employee retention. The organisation should establish clear career pathways, provide regular training programmes, implement mentoring initiatives, and create opportunities for internal promotions to improve employee engagement and long-term retention.

4. `Review Compensation and Reward Policies`

The analysis suggests that employees with lower monthly incomes may be more likely to leave the organisation. Management should periodically benchmark salaries against industry standards, review compensation packages, introduce performance-based incentives, and recognise outstanding employee contributions to improve job satisfaction and reduce voluntary turnover.

5. `Adopt Predictive HR Analytics for Proactive Decision-Making`

The exploratory analysis identified variables such as overtime, department, job role, age, and monthly income as potential predictors of employee attrition. The organisation should implement predictive analytics and machine learning models to identify employees who may be at risk of leaving. This will enable Human Resource managers to implement proactive retention strategies, optimise workforce planning, and make evidence-based decisions that improve organisational performance.
